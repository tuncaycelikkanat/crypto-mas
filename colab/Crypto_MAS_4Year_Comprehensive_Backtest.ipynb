## 3. Sonuçların Görselleştirilmesi ve Analizi (Analytics & Charts)
4.5-Yıllık WFO süitinin ürettiği JSON raporlarını yükleyip sermaye eğrisi (Equity Curve), kazanç tutarlılık oranı (Consistency Ratio) ve sembol havuzu kırılımlarını çizdiriyoruz.

```python
# @title 3.1 Aylık ve Haftalık WFO Performans Matrisi
def plot_wfo_summary(file_path, title):
    if not os.path.exists(file_path):
        print(f"⚠️ Dosya bulunamadı: {file_path}. Lütfen önce 2.1 adımını çalıştırın.")
        return
    with open(file_path, 'r', encoding='utf-8') as f:
        data = json.load(f)
    
    results = data.get('results', [])
    df = pd.DataFrame(results)
    
    print(f"\n{'='*60}")
    print(f"📊 {title.upper()} - 4.5 YILLIK PERFORMANS ÖZETİ (2022-01-01 -> 2026-07-01)")
    print(f"{'='*60}")
    display(df[['pool_name', 'total_4_year_profit_usd', 'total_trades', 'consistency_ratio_pct', 'total_periods_evaluated']])
    
    # Çubuk Grafik
    plt.figure(figsize=(10, 5))
    sns.barplot(data=df, x='pool_name', y='total_4_year_profit_usd', palette='viridis')
    plt.title(f"{title} — Havuz Bazlı 4.5-Yıllık Net Kâr (USD)")
    plt.ylabel("Net Kâr (USD)")
    plt.xlabel("Varlık Havuzu")
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.show()

plot_wfo_summary("colab/results/4year_monthly_wfo.json", "Aylık Granüler WFO")
plot_wfo_summary("colab/results/4year_weekly_wfo.json", "Haftalık Granüler WFO")
```
